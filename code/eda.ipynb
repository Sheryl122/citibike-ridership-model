```python
# E4
# Extract months and years
x = pd.to_datetime((df['ride_date'])) # filter on weather
y = df['num_rides']

plt.plot(x,y)
plt.show()
#Isolate the days that have the same weather
```

```python
df['ride_date'] = pd.to_datetime(df['ride_date'])

plt.plot(df['ride_date'], df['num_rides'])
df['year'] = df['ride_date'].dt.year
df['month'] = df['ride_date'].dt.month

december_by_year = df[df['month'] == 12].groupby('year')['num_rides'].mean()
print(december_by_year)
```

```python
december_by_year.plot(kind='bar')
```

```python
# E5
week_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

avg_by_day = df.groupby('day_of_week')['num_rides'].mean().reindex(week_order)

print(avg_by_day.round(2))

avg_by_day.plot(kind='bar', color='purple')
plt.title('Average Rideship By Day of Week')
plt.xlabel('Day Of Week')
plt.ylabel('Average RiderShip')
```
