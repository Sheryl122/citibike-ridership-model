```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = pd.read_csv('../data/citibike_weather_daily.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
x = df['precip_in']
y = df['num_rides']
plt.scatter(x,y)
plt.xlabel('Precipitation')
plt.ylabel('Num_Rides')
plt.title('Num_rides vs Rain')
plt.show()
```

```python
from turtle import color

x = df['temp_f']
y = df['num_rides']


plt.title('Ride Count vs Temperature')

colors = ['purple' if val > 55 else 'm' for val in x]
plt.scatter(x,y, color= colors)
plt.xlabel('Temperature')
plt.ylabel('Number Of Rides')
plt.show()
```

```python
x = df['wind_speed_knots']
y = df['num_rides']
plt.scatter(x,y)
```

### Number of Rides VS Weather Variable
Based on these charts it is clear that when the weather is hot that is when ridership rises. As well as when there is zero to minimal rain. Rideship seems to slow down when wind speed knots rises but not as much as with rain.
The strongest relationship is ride_num against precipitation and the weakest is against wind_speed_knot.

```python
df['ride_date'] = pd.to_datetime(df['ride_date'])

df['year'] = df['ride_date'].dt.year
df['month'] = df['ride_date'].dt.month

december_by_year = df[df['month'] == 12].groupby('year')['num_rides'].mean().round(2)
print(december_by_year)
```

```python
december_by_year.plot(kind='bar', color='m')
plt.title('Rideship Count Each December')
plt.ylabel('Rideship Count')
```

### E4
I decided to compare the month December across all the years. I noticed that 2013 December is actually more than Decmber 2014. The amount of ridership really increase from the year 2015 to 2017.

```python
# E5
week_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

avg_by_day = df.groupby('day_of_week')['num_rides'].mean().reindex(week_order)

print(avg_by_day.round(2))

avg_by_day.plot(kind='bar', color='purple')
plt.title('Average Rideship By Day of Week')
plt.xlabel('Day Of Week')
plt.ylabel('Average RiderShip')
```

# Rideship Daily Observation
Based on the chart the rideship average is higher on weekdays, whilee the weekend is significantly lower. Thepattern does match my intuiton about who rides Citi Bike, which is most likely workers using the bikes to commute. This would also explain why. the ridership average lowers during the weekend becasue most people don't work on the weekends.


```python
# E6
range_by_date = pd.date_range(start=df['ride_date'].min(), end=df['ride_date'].max())
print(len(range_by_date))
print(len(df))

missing_days = len(range_by_date) - len(df)
print(missing_days)
```

```python
missing_dates = []
for date in range_by_date:
    if date not in df['ride_date'].values:
        missing_dates.append(date)

print(missing_dates)
```

```python
missing_df = pd.DataFrame({'missing_dates': missing_dates})
missing_df.describe()
```

```python
missing_df.head()
```

```python
missing_df['year'] = missing_df['missing_dates'].dt.year
missing_df['month'] = missing_df['missing_dates'].dt.month
print(missing_df['year'])
print(missing_df['month'])
```

```python
missing_df.pivot_table(index='year', columns='month', values='missing_dates', aggfunc='count', fill_value=0)
```
