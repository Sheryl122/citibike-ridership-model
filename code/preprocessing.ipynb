```python
day_of_week_dummies = pd.get_dummies(df['day_of_week'], drop_first=True)
print(day_of_week_dummies)
```

## Reason for drop_first
I decided to use the drop_first because it prevents multicollinearity. Which would cause problems down the line during the model creation.

```python
df = pd.concat([df, day_of_week_dummies], axis=1)
```
